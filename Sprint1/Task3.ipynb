Вариант задания № 2
Описание задачи:
Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Student:
• Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
• Методы:
o GetInfo(): метод для получения информации о студенте в виде строки.
o Study(): метод для вывода сообщения о том, что студент учится.
o TakeExam(): метод для вывода сообщения о сдаче экзамена.
Требования к производным классам:
1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
переопределен для добавления информации о специальности при изучении
предмета.
2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
атрибуты, такие как Научный руководитель (ScientificAdvisor).
Метод TakeExam() должен быть переопределен для добавления
информации о научном руководителе при сдаче экзамена.
3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
Метод Study() должен быть переопределен для добавления информации о
теме диссертации при изучении предмета.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

```
public interface IStudentID
{
    int ID {get; set;}    
}

public interface IBachelor
{
    int NumberCourses {get; set;}
    bool StudiesOnSaturday {get; set;}
    void IsFullTime();
}

public interface IPhD
{
    void PublishPaper();
    void AttendConferences();
}

public interface IMaster
{
    string SubjectTaught {get; set;}
    void GradeAssignments();
}



public class Student
{
    private string name;
    public string Name 
    {
        get {return name;}
        set {name = value;}
    }

    private int age;
    public int Age 
    {
        get {return age;}
        set {age = value;}
    }

    private int course;
    public int Course 
    {
        get {return course;}
        set {course = value;}
    }

    private static int totalStudent;
    public static int TotalStudent
    {
        get {return totalStudent;}
        private set { totalStudent = value;}
    }

    public Student() 
    {
        totalStudent++;
    }

    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
        totalStudent++;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine("Basic student information: " + Name + ", " + Age + " years old " + ", " + Course + "- year student");
    }

    public virtual void Study()
    {
        Console.WriteLine("The student is studying");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine("The student is taking an exam");
    }

    public virtual void StudyTogether(Student student2)
    {
        Console.WriteLine(Name + " studies with " + student2.Name + " in the library");
    }

    public virtual void Interact(Student student2)
    {
        Console.WriteLine(Name + " learns with " + student2.Name);
    }

    public static void DisplayTotalStudent()
    {
        Console.WriteLine("Total Student is " + TotalStudent);
    }

}




public class BachelorStudent : Student, IStudentID, IBachelor
{
    private string speciality;
    public string Speciality     
    {
        get { return speciality; }
        set { speciality = value; }
    }
    public int ID {get; set;} 
    public int NumberCourses {get; set;}  
    public bool StudiesOnSaturday {get; set;} 

    public BachelorStudent() : base()
    {

    }

    public BachelorStudent(string name, int age, int course, string speciality, int iD, int numberCourses, bool studiesOnSaturday) : base(name, age, course)
    {
        Name = name;
        Age = age;
        Course = course;
        Speciality = speciality;
        ID = iD;
        NumberCourses = numberCourses;
        StudiesOnSaturday = studiesOnSaturday;
    }

    public override void Study()
    {
        Console.WriteLine("The bachelor's degree program of the student is " + Speciality);
    }

    public override void GetInfo()
    {
        Console.WriteLine("Bachelor Student information: " + Name + ", " + Age + " years old " + ", " + Course + "- year student, " + "Speciality: " + Speciality + ", ID: " + ID +", Number of Courses: " + NumberCourses + ", Studies on Saturday: "+ StudiesOnSaturday);
    }

    public override void Interact(Student student2)
    {
        Console.WriteLine(Name + " learns with " + student2.Name + " " + Speciality);
    }

    public void IsFullTime()
    {
        if (NumberCourses > 5)
        {
            Console.WriteLine($"{Name} is a full time student");
        }        
        else
        {
            Console.WriteLine($"{Name} is not a full time student");
        }        
    }
}

public class MasterStudent : Student, IStudentID
{
    private string scientificAdvisor;
    public string ScientificAdvisor 
    {
        get { return scientificAdvisor; }
        set { scientificAdvisor = value; }
    }
    public int ID {get; set;}
    public string SubjectTaught {get; set;} 

    public MasterStudent() : base()
    {

    }

    public MasterStudent(string name, int age, int course, string scientificAdvisor, int iD, string subjectTaught) : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
        ID = iD;
        SubjectTaught = subjectTaught;
    }

    public override void TakeExam()
    {
        Console.WriteLine("The master student is taking the exam with " + ScientificAdvisor);
    }

    public override void GetInfo()
    {
        Console.WriteLine("Master Student information: " + Name + ", " + Age + " years old " + ", " + Course + "- year student," + " ID: " + ID + ", Teaching assistant for " + SubjectTaught);
    }

    public override void Interact(Student student2)
    {
        Console.WriteLine(Name + " learns research from " + student2.Name + " under the supervision of " + ScientificAdvisor);
    }

    public void GradeAssignments()
    {
        Console.WriteLine($"{Name} is grading assignments for the course {SubjectTaught}");
    }
}


public class PhDStudent : Student, IStudentID, IPhD
{
    private string dissertationTopic;
    public string DissertationTopic 
    {
        get { return dissertationTopic; }
        set { dissertationTopic = value; }
    }
    public int ID {get; set;} 

    public PhDStudent() : base()
    {}

    public PhDStudent(string name, int age, int course, string dissertationTopic, int iD) : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
        ID = iD;
    }

    public override void Study()
    {
        Console.WriteLine("The PhD student is studying a subject related to " + DissertationTopic);
    }

    public override void GetInfo()
    {
        Console.WriteLine("PhD Student information: " + Name + ", " + Age + " years old " + ", " + Course + "- year student" + ", ID: " + ID);
    }

    public override void Interact(Student student2)
    {
        Console.WriteLine(Name + " gives " + student2.Name + " advice about " + DissertationTopic);
    }

    public void PublishPaper()
    {
        Console.WriteLine($"{Name} has published two research papers on {DissertationTopic} this year");
    }

    public void AttendConferences()
    {
        Console.WriteLine($"{Name} is required to attend three conferences on {DissertationTopic} at least this year");
    }
}

public class InternationBachelorStudent : BachelorStudent
{
    public int VisaNumber {get; set;}
    public string CountryOfOrigin {get; set;}
    
    
    public InternationBachelorStudent(string name, int age, int course, string speciality, int iD, int numberCourses, bool studiesOnSaturday, int visaNumber, string countryOfOrigin) : base(name, age, course, speciality, iD, numberCourses, studiesOnSaturday)
    {
        VisaNumber = visaNumber;
        CountryOfOrigin = countryOfOrigin;
        
    }

    public override void StudyTogether(Student student3)
    {
      // así también puede ser, recuerda!!!  Console.WriteLine($"{Name} studies {Speciality} with {student3.Name}");
        Console.WriteLine(Name + " studies " + Speciality +" with " +  student3.Name);
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Visa Number: {VisaNumber}, Country of Origin: {CountryOfOrigin}");
    }
}


Student myStudent = new Student();
BachelorStudent myBachelorStudent = new BachelorStudent("Fabrizzio", 18, 1, "Accounting", 2026001, 9, false);
MasterStudent myMasterStudent = new MasterStudent("Gabriela", 26, 2, "Dr. House", 2026032, "Machine Learning");
PhDStudent myPhDStudent = new PhDStudent("Daniela", 30, 3, "Artificial Intelligence", 20250011);
InternationBachelorStudent myInternationBachelorStudent = new InternationBachelorStudent("Juan", 27, 2, "Programming", 2026025, 10, true, 123456789, "Perú");

Console.WriteLine("----------Bachelor Student's Info----------");
myBachelorStudent.GetInfo();
myBachelorStudent.Study();
myBachelorStudent.IsFullTime();

Console.WriteLine("----------International Bachelor Student's Info----------");
myInternationBachelorStudent.GetInfo();
myInternationBachelorStudent.IsFullTime();


Console.WriteLine("----------Master Student's Info----------");
myMasterStudent.GetInfo();
myMasterStudent.TakeExam();

Console.WriteLine("----------PhD Student's Info----------");
myPhDStudent.GetInfo();
myPhDStudent.Study();

Console.WriteLine("----------Objects Interaction Info----------");
myBachelorStudent.StudyTogether(myMasterStudent);
myBachelorStudent.Interact(myMasterStudent);
myMasterStudent.Interact(myPhDStudent);
myPhDStudent.Interact(myMasterStudent);
myInternationBachelorStudent.StudyTogether(myBachelorStudent);
myMasterStudent.GradeAssignments();
myPhDStudent.AttendConferences();
myPhDStudent.PublishPaper();

Student.DisplayTotalStudent();



```
