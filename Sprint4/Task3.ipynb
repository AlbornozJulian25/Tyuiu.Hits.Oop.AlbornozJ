Вариант задания № 2
Описание задачи:
Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Student:
• Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
• Методы:
o GetInfo(): метод для получения информации о студенте в виде строки.
o Study(): метод для вывода сообщения о том, что студент учится.
o TakeExam(): метод для вывода сообщения о сдаче экзамена.
Требования к производным классам:
1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
переопределен для добавления информации о специальности при изучении
предмета.
2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
атрибуты, такие как Научный руководитель (ScientificAdvisor).
Метод TakeExam() должен быть переопределен для добавления
информации о научном руководителе при сдаче экзамена.
3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
Метод Study() должен быть переопределен для добавления информации о
теме диссертации при изучении предмета.

#### Дополнительное задание

```
public interface IStudent 
{
    void GetInfo();
    void TakeExam();
    void Study();
}



public interface IStudentID
{
    int ID {get; set;}    
}

public interface IBachelor
{
    int NumberCourses {get; set;}
    bool StudiesOnSaturday {get; set;}
    void IsFullTime();
}

public interface IPhD
{
    void PublishPaper();
    void AttendConferences();
}

public interface IMaster
{
    string SubjectTaught {get; set;}
    void GradeAssignments();
}


public class Student : IStudent
{
    private string name;
    public string Name 
    {
        get {return name;}
        set {name = value;}
    }

    private int age;
    public int Age 
    {
        get {return age;}
        set {age = value;}
    }

    private int course;
    public int Course 
    {
        get {return course;}
        set {course = value;}
    }

    private static int totalStudent;
    public static int TotalStudent
    {
        get {return totalStudent;}
        private set { totalStudent = value;}
    }

    public Student() 
    {
        totalStudent++;
    }

    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
        totalStudent++;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine("Basic student information: " + Name + ", " + Age + " years old " + ", " + Course + "- year student");
    }

    public virtual void Study()
    {
        Console.WriteLine("The student is studying");
    }


    public virtual void Study(int hours)
    {
        Console.WriteLine($"{Name} studies many {hours} a day");
    }

    public virtual void Study(string place)
    {
        Console.WriteLine($"{Name} studies in {place}");
    }


    public virtual void TakeExam()
    {
        Console.WriteLine("The student is taking an exam");
    }

    public virtual void StudyTogether(Student student2)
    {
        Console.WriteLine(Name + " studies with " + student2.Name + " in the library");
    }

    public virtual void Interact(Student student2)
    {
        Console.WriteLine(Name + " learns with " + student2.Name);
    }


    public virtual void Interact(Student student2, string location)
    {
        Console.WriteLine($"{Name} learns with {student2.Name} in {location} after classes");
    }

    public virtual void Interact(Student student2, int amount)
    {
        Console.WriteLine($"{Name} has professional conversations with {student2.Name} {amount} times a month");
    }

    void IStudent.GetInfo()
    {
        Console.WriteLine($"Basic student information: |Name = {Name}| Age: {Age}| Course: {Course} - yeard course");
    }

    void IStudent.TakeExam()
    {
        Console.WriteLine($"{Name} is taking an exam");
    }

    void IStudent.Study()
    {
        Console.WriteLine($"{Name} is styuding");

    }

    public static void DisplayTotalStudent()
    {
        Console.WriteLine("Total Student is " + TotalStudent);
    }

}

public class BachelorStudent : Student, IStudentID, IBachelor, IStudent
{
    private string speciality;
    public string Speciality     
    {
        get { return speciality; }
        set { speciality = value; }
    }
    public int ID {get; set;} 
    public int NumberCourses {get; set;}  
    public bool StudiesOnSaturday {get; set;} 

    public BachelorStudent() : base()
    {

    }

    public BachelorStudent(string name, int age, int course, string speciality, int iD, int numberCourses, bool studiesOnSaturday) : base(name, age, course)
    {
        Name = name;
        Age = age;
        Course = course;
        Speciality = speciality;
        ID = iD;
        NumberCourses = numberCourses;
        StudiesOnSaturday = studiesOnSaturday;
    }

    public override void Study()
    {
        Console.WriteLine("The bachelor's degree program of the student is " + Speciality);
    }


    public override void Study(int hours)
    {
        Console.WriteLine($"{Name} studies {Speciality} {hours} hours everyday during the week");
    }

    public override void Study(string place)
    {
        Console.WriteLine($"{Name} likes to study at {place} in the mornings");
    }


    public override void GetInfo()
    {
        Console.WriteLine("Bachelor Student information: " + Name + ", " + Age + " years old " + ", " + Course + "- year student, " + "Speciality: " + Speciality + ", ID: " + ID +", Number of Courses: " + NumberCourses + ", Studies on Saturday: "+ StudiesOnSaturday);
    }

    public override void Interact(Student student2)
    {
        Console.WriteLine(Name + " learns with " + student2.Name + " " + Speciality);
    }


    public override void Interact(Student student2, string location)
    {
        Console.WriteLine($"{Name} prefers to learn with {student2.Name} at {location} due to convenience");
    }

    public override void Interact(Student student2, int amount)
    {
        Console.WriteLine($"{Name} has always debates with {student2.Name} and other students about {Speciality} for {amount} hours every friday night");
    }


    public void IsFullTime()
    {
        if (NumberCourses > 5)
        {
            Console.WriteLine($"{Name} is a full time student");
        }        
        else
        {
            Console.WriteLine($"{Name} is not a full time student");
        }        
    }

    void IStudent.GetInfo()
    {
        Console.WriteLine($"Bachelor Student information:: |Name = {Name}| Age: {Age}| Course: {Course} - yeard course| Speciality: {Speciality}| ID: {ID}| Number of Courses: {NumberCourses}| Studies on Saturday: {StudiesOnSaturday}");
    }

    void IStudent.TakeExam()
    {
        Console.WriteLine($"{Name} is taking a {Speciality} exam next week");
    }

    void IStudent.Study()
    {
        Console.WriteLine($"{Name}'s Bachelor degree program is {Speciality}");

    }
}

public class MasterStudent : Student, IStudentID, IStudent
{
    private string scientificAdvisor;
    public string ScientificAdvisor 
    {
        get { return scientificAdvisor; }
        set { scientificAdvisor = value; }
    }
    public int ID {get; set;}
    public string SubjectTaught {get; set;} 

    public MasterStudent() : base()
    {

    }

    public MasterStudent(string name, int age, int course, string scientificAdvisor, int iD, string subjectTaught) : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
        ID = iD;
        SubjectTaught = subjectTaught;
    }

    public override void TakeExam()
    {
        Console.WriteLine("The master student is taking the exam with " + ScientificAdvisor);
    }

    public override void GetInfo()
    {
        Console.WriteLine("Master Student information: " + Name + ", " + Age + " years old " + ", " + Course + "- year student," + " ID: " + ID + ", Teaching assistant for " + SubjectTaught);
    }

    public override void Interact(Student student2)
    {
        Console.WriteLine(Name + " learns research from " + student2.Name + " under the supervision of " + ScientificAdvisor);
    }

    public void GradeAssignments()
    {
        Console.WriteLine($"{Name} is grading assignments for the course {SubjectTaught}");
    }

    void IStudent.GetInfo()
    {
        Console.WriteLine($"Master Student information: |Name = {Name}| Age: {Age}| Course: {Course} - yeard course| Scientific Advisor: {ScientificAdvisor}| ID: {ID}| Subject Taught: {SubjectTaught} ");
    }

    void IStudent.TakeExam()
    {
        Console.WriteLine($"{Name} is taking the exam with {ScientificAdvisor}");
    }

    void IStudent.Study()
    {
        Console.WriteLine($"{Name} is styuding all nights for the {SubjectTaught} exam");
    }
}

public class PhDStudent : Student, IStudentID, IPhD, IStudent
{
    private string dissertationTopic;
    public string DissertationTopic 
    {
        get { return dissertationTopic; }
        set { dissertationTopic = value; }
    }
    public int ID {get; set;} 

    public PhDStudent() : base()
    {
        
    }

    public PhDStudent(string name, int age, int course, string dissertationTopic, int iD) : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
        ID = iD;
    }

    public override void Study()
    {
        Console.WriteLine("The PhD student is studying a subject related to " + DissertationTopic);
    }


    public override void Study(int hours)
    {
        Console.WriteLine($"{Name} studies about {DissertationTopic} {hours} hours every weekend");
    }

    public override void Study(string place)
    {
        Console.WriteLine($"{Name} studies better and get more relevant information about {DissertationTopic} in the {place}");
    }


    public override void GetInfo()
    {
        Console.WriteLine("PhD Student information: " + Name + ", " + Age + " years old " + ", " + Course + "- year student, " + DissertationTopic + ", ID: " + ID);
    }

    public override void Interact(Student student2)
    {
        Console.WriteLine(Name + " gives " + student2.Name + " advice about " + DissertationTopic);
    }


    public override void Interact(Student student2, string location)
    {
        Console.WriteLine($"{Name} gives private classes to {student2.Name} in {location}");        
    }

    public override void Interact(Student student2, int amount)
    {
        Console.WriteLine($"{Name} provides academic guidance to {student2.Name} before exams {amount} hours a week");
    }


    public void PublishPaper()
    {
        Console.WriteLine($"{Name} has published two research papers on {DissertationTopic} this year");
    }

    public void AttendConferences()
    {
        Console.WriteLine($"{Name} is required to attend three conferences on {DissertationTopic} at least this year");
    }

    void IStudent.GetInfo()
    {
        Console.WriteLine($"PhD Student information: |Name = {Name}| Age: {Age}| Course: {Course} - yeard course| Dissertation Topic: {DissertationTopic}| ID: {ID}");
    }

    void IStudent.TakeExam()
    {
        Console.WriteLine($"{Name} is taking the exam about {DissertationTopic} tomorrow morning");
    }

    void IStudent.Study()
    {
        Console.WriteLine($"{Name} is styuding a subject related to {DissertationTopic}");

    }
}

public class InternationBachelorStudent : BachelorStudent, IStudent
{
    public int VisaNumber {get; set;}
    public string CountryOfOrigin {get; set;}


    public InternationBachelorStudent() : base() 
    {

    }    
    
    public InternationBachelorStudent(string name, int age, int course, string speciality, int iD, int numberCourses, bool studiesOnSaturday, int visaNumber, string countryOfOrigin) : base(name, age, course, speciality, iD, numberCourses, studiesOnSaturday)
    {
        VisaNumber = visaNumber;
        CountryOfOrigin = countryOfOrigin;
        
    }

    public override void StudyTogether(Student student3)
    {
      // así también puede ser, recuerda!!!  Console.WriteLine($"{Name} studies {Speciality} with {student3.Name}");
        Console.WriteLine(Name + " studies " + Speciality +" with " +  student3.Name);
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Visa Number: {VisaNumber}, Country of Origin: {CountryOfOrigin}");
    }




    void IStudent.GetInfo()
    {
        Console.WriteLine($"International Bachelor Student information: |Name: {Name}| Age: {Age}| Course: {Course} - yeard course| Speciality: {Speciality}| ID: {ID}| Number of Courses: {NumberCourses}| Studies on saturday: {StudiesOnSaturday}| Visa Number: {VisaNumber}| Country of origin: {CountryOfOrigin}");
    }

    void IStudent.TakeExam()
    {
        Console.WriteLine($"{Name} is taking the exam about {Speciality} tomorrow night");
    }

    void IStudent.Study()
    {
        Console.WriteLine($"{Name} is styuding a subject related to {Speciality}");

    }
}


public class StudentRegistration<T> where T : Student
{
    private List<T> students = new List<T>();

    public void Add(T student)
    {
        students.Add(student);
    }

    public void Remove(T student)
    {
        students.Remove(student);
    }

    public void ShowStudentRegistration()
    {
        foreach(var student in students)
        {
            student.GetInfo();
        }
    }
}

public class StudentService
{
    private readonly  IStudent _student;

    public StudentService(IStudent student)
    {
        _student = student;
    }

    public void MethodsForStudent()
    {
        _student.GetInfo();
        _student.TakeExam( );
        _student.Study();
    }
}


Student myStudent = new Student();
BachelorStudent myBachelorStudent = new BachelorStudent("Fabrizzio", 18, 1, "Accounting", 2026001, 9, false);
MasterStudent myMasterStudent = new MasterStudent("Gabriela", 26, 2, "Dr. House", 2026032, "Machine Learning");
PhDStudent myPhDStudent = new PhDStudent("Daniela", 30, 3, "Artificial Intelligence", 2025011);
InternationBachelorStudent myInternationBachelorStudent = new InternationBachelorStudent("Juan", 27, 2, "Programming", 2026025, 10, true, 123456789, "Perú");

Console.WriteLine("----------Bachelor Student's Info----------");
myBachelorStudent.GetInfo();
myBachelorStudent.Study();
myBachelorStudent.IsFullTime();
myBachelorStudent.Study(5);
myBachelorStudent.Study("university's library");
myBachelorStudent.Interact(myMasterStudent, "home");
myBachelorStudent.Interact(myInternationBachelorStudent, 2);


Console.WriteLine("----------International Bachelor Student's Info----------");
myInternationBachelorStudent.GetInfo();
myInternationBachelorStudent.IsFullTime();


Console.WriteLine("----------Master Student's Info----------");
myMasterStudent.GetInfo();
myMasterStudent.TakeExam();

Console.WriteLine("----------PhD Student's Info----------");
myPhDStudent.GetInfo();
myPhDStudent.Study();
myPhDStudent.Study(10);
myPhDStudent.Study("The National Library");
myPhDStudent.Interact(myMasterStudent, "Starbucks");
myPhDStudent.Interact(myMasterStudent, 6);

Console.WriteLine("----------Objects Interaction Info----------");
myBachelorStudent.StudyTogether(myMasterStudent);
myBachelorStudent.Interact(myMasterStudent);
myMasterStudent.Interact(myPhDStudent);
myPhDStudent.Interact(myMasterStudent);
myInternationBachelorStudent.StudyTogether(myBachelorStudent);
myMasterStudent.GradeAssignments();
myPhDStudent.AttendConferences();
myPhDStudent.PublishPaper();

Console.WriteLine("----------Student Registration's Info (Using Generic Class)----------");
StudentRegistration<Student> students = new StudentRegistration<Student>();
students.Add(myBachelorStudent);
students.Add(myMasterStudent);
students.Add(myPhDStudent);
students.Add(myInternationBachelorStudent);
students.ShowStudentRegistration();


Console.WriteLine("----------Explicit Interface Implementation----------");
IStudent myStudent2 = new Student { Name = "Lolo", Age = 19, Course = 3 };
myStudent2.GetInfo();
myStudent2.TakeExam();
myStudent2.Study();


IStudent myBachelorStudent2 = new BachelorStudent {  Name = "Ricardo", Age= 23, Course = 4, Speciality = "Finances", ID = 2026567, NumberCourses = 8, StudiesOnSaturday = true }; 
myBachelorStudent2.GetInfo();
myBachelorStudent2.TakeExam();
myBachelorStudent2.Study();


IStudent myMasterStudent2 = new MasterStudent { Name = "Macarena", Age = 27, Course = 1, ScientificAdvisor = "Dr. Dolittle", ID = 2026781, SubjectTaught = "Advanced SQL"};
myMasterStudent2.GetInfo();
myMasterStudent2.TakeExam();
myMasterStudent2.Study();


IStudent myPhDStudent2 = new PhDStudent { Name = "Yolanda", Age = 37, Course = 3, DissertationTopic = "Advanced AI", ID = 2026222 };
myPhDStudent2.GetInfo();
myPhDStudent2.TakeExam();
myPhDStudent2.Study();


IStudent myInternationBachelorStudent2 = new InternationBachelorStudent { Name = "Cook", Age = 17, Course = 1, Speciality = "Programming at Oil Enterprices", ID = 2026154, NumberCourses = 10, StudiesOnSaturday = true, VisaNumber = 123456745, CountryOfOrigin = "Brazil" };
myInternationBachelorStudent2.GetInfo();
myInternationBachelorStudent2.TakeExam();
myInternationBachelorStudent2.Study();


Console.WriteLine("----------Dependency Injection----------");
IStudent myBachelorStudent3 = new BachelorStudent("Cristiano", 21, 2, "Management", 2026014, 8, true);
StudentService bachelorStudentService = new StudentService(myBachelorStudent3);
bachelorStudentService.MethodsForStudent();

IStudent myMasterStudent3 = new MasterStudent("Claudia", 24, 1, "Dr. García", 2026259, "Data Science");
StudentService masterStudentService = new StudentService(myMasterStudent3);
masterStudentService.MethodsForStudent();

IStudent myPhDStudent3 = new PhDStudent("Danize", 47, 2, "Data Analysis", 2025047);
StudentService phDStudentService = new StudentService(myPhDStudent3);
phDStudentService.MethodsForStudent();

IStudent myInternationBachelorStudent3 = new InternationBachelorStudent("Earling", 29, 4, "Programming and AI", 2026555, 9, false, 123456754, "Norway");
StudentService internationBachelorStudentService = new StudentService(myInternationBachelorStudent3);
internationBachelorStudentService.MethodsForStudent();

Student.DisplayTotalStudent();
```
