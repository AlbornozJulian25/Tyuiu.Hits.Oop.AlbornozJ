```polyglot-notebook
int [] numbers = {5, 8, 1, 3, 9, 6, 7, 2};

int сумма = numbers.Sum();

Console.WriteLine("Сумма всех элементов массива: " + сумма);

```

```polyglot-notebook
int [] numbers = {5, 8, 1, 3, 9, 6, 7, 2};

double Среднее = numbers.Average();

Console.WriteLine("Среднее арифметическое: " + Среднее);
```

```polyglot-notebook
int [] numbers = {5, 8, 1, 3, 9, 6, 7, 2};

int Минимум = numbers.Min();
int Максимум = numbers.Max();

Console.WriteLine("Минимальное число в массиве: " + Минимум);
Console.WriteLine("Максимальное число в массиве: " + Максимум);

```

```polyglot-notebook
int [] numbers = {5, 8, 1, 3, 9, 6, 7, 2};

int [] Отсортирован = numbers.OrderBy(n=>n).ToArray();
Console.WriteLine("Отсортированный массив: " + string.Join(", ", Отсортирован));
```

```polyglot-notebook
int [] numbers = {5, 8, 1, 3, 9, 6, 7, 2};

int numbertofind = 7;

bool contains = numbers.Contains(numbertofind);

Console.WriteLine("Содержит ли массив число: " + numbertofind + " ? " + contains);

```
